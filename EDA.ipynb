# Loading the Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("ecommerce_sales_dataset(1).xls")
```

# Initial Exploration of the Dataset

```python
df.head()
```

```python
df.describe()
```

# Handling Missing Values

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

### Observation

No missing values were found in the dataset.

Therefore, no missing value treatment was required.

# Removing Duplicate Records

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

### Observation

Duplicate records were checked and removed if present.

The dataset now contains only unique records.

### Observation

No duplicate records were found in the dataset.

Therefore, no duplicate rows needed to be removed.

# Checking and Correcting Data Types

```python
df.dtypes
```

```python
df.select_dtypes(include=['object']).columns
```

# Correcting Data Types

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df.dtypes
```

### Observation

The `Order_Date` column was originally stored as a string.

It was converted to datetime format to enable date-based analysis and visualization.

# Correlation Heatmap

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

The correlation heatmap was created to understand the relationships between numerical variables in the dataset. Strong positive correlations are close to +1, strong negative correlations are close to -1, while values near 0 indicate weak or no relationship.

# Distribution Plot of Revenue

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Revenue"], bins=30, kde=True)

plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

### Observation

This distribution plot shows how revenue values are distributed across the dataset. It helps identify whether most sales have low, medium, or high revenue.

# Revenue by Category

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Category",
    y="Revenue",
    estimator="mean"
)

plt.title("Average Revenue by Category")
plt.xticks(rotation=45)

plt.show()
```

### Observation

This chart compares the average revenue generated by each product category.

# Revenue by Region

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x="Region",
    y="Revenue",
    estimator="mean"
)

plt.title("Average Revenue by Region")
plt.xticks(rotation=45)

plt.show()
```

### Observation

This chart compares the average revenue generated across different regions.

# Profit by Customer Segment

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Customer_Segment",
    y="Profit",
    estimator="mean"
)

plt.title("Average Profit by Customer Segment")

plt.show()
```

### Observation

This chart shows which customer segment generates the highest average profit.

# Shipping Analysis

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="Shipping_Method",
    y="Shipping_Cost",
    estimator="mean"
)

plt.title("Average Shipping Cost by Shipping Method")
plt.xticks(rotation=20)

plt.show()
```

### Observation

This chart compares the average shipping cost for different shipping methods.

# Order Status Analysis

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Order_Status")

plt.title("Order Status Distribution")

plt.show()
```

### Observation

This chart shows the number of orders in each order status category.

# Business Insights

### 1. Revenue and Cost have a very strong positive correlation.
The correlation heatmap shows that Revenue and Cost are highly correlated (0.96), indicating that as revenue increases, the overall cost also increases.

### 2. Electronics category generates the highest average revenue.
Among all product categories, Electronics has the highest average revenue, while Health generates the lowest average revenue.

### 3. Middle East region has the highest average revenue.
The regional analysis shows that the Middle East has the highest average revenue, whereas North America has the lowest average revenue among the four regions.

### 4. New customers generate slightly higher average profit.
The customer segment analysis indicates that New customers have the highest average profit, while Premium customers have the lowest average profit.

### 5. Overnight shipping is the most expensive shipping method.
The shipping analysis shows that Overnight shipping has the highest average shipping cost, whereas Economy shipping has the lowest shipping cost.

# Conclusion

The exploratory data analysis revealed important business patterns in the Global E-Commerce Sales dataset. Electronics and the Middle East region generated the highest average revenue, while Revenue showed a strong positive relationship with Cost and Profit. Overnight shipping had the highest shipping cost, and most orders were successfully delivered. These insights can help businesses improve pricing strategies, optimize shipping methods, and focus on high-performing categories and regions.
